```python
Q1. Phone Contact Book
--------------------------------------------------------------
Create a dictionary to store contacts (name -> phone number).
Show a small menu in a while loop:
  1. Add contact
  2. Search contact
  3. Show all contacts
  4. Exit

  - While adding, if the name already exists, print
    "Contact already exists".
  - While searching, if the name is not found, print
    "Contact not found".
```

```python
contacts = {}

while True:

    print("\n--- Phone Contact Book ---")
    print("1. Add contact")
    print("2. Search contact")
    print("3. Show all contacts")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":

        name = input("Enter name: ")
        phone = input("Enter phone number: ")

        if name in contacts:
            print("Contact already exists")
        else:
            contacts[name] = phone
            print("Contact added successfully")

    elif choice == "2":

        name = input("Enter name to search: ")

        if name in contacts:
            print("Phone number:", contacts[name])
        else:
            print("Contact not found")

    elif choice == "3":

        if len(contacts) == 0:
            print("No contacts available")
        else:
            print("\nContacts:")

            for name, phone in contacts.items():
                print(name, ":", phone)

    elif choice == "4":

        print("Goodbye!")
        break

    else:
        print("Invalid choice")
```

```python
Q2. Unique Visitors on a Website
--------------------------------------------------------------
A website logged visitor IDs throughout the day (with repeats):
  visits = [101, 205, 101, 330, 205, 412, 101, 330, 509]

Using a set, find and print:
  - The number of total visits
  - The number of unique visitors
  - The unique visitor IDs in sorted order
```

```python
visits = [101, 205, 101, 330, 205, 412, 101, 330, 509]

unique_visitors = set(visits)

print("Total visits:", len(visits))
print("Number of unique visitors:", len(unique_visitors))
print("Unique visitor IDs:", sorted(unique_visitors))
```

```python
Q3. Student Marks Dictionary
--------------------------------------------------------------
Marks of students are stored in a dictionary:
  marks = {"Aman": 82, "Priya": 91, "Rahul": 45,
           "Sneha": 67, "Kiran": 30}

Write a program to:
  - Print each student's name with "Pass" or "Fail"
    (pass mark is 40)
  - Find the topper and their marks
  - Create a new dictionary with only the students who passed
```

```python
marks = {
    "Aman": 82,
    "Priya": 91,
    "Rahul": 45,
    "Sneha": 67,
    "Kiran": 30
}

passed_students = {}

topper_name = ""
topper_marks = 0

for name, mark in marks.items():

    if mark >= 40:
        print(name, "-", mark, "- Pass")
        passed_students[name] = mark
    else:
        print(name, "-", mark, "- Fail")

    if mark > topper_marks:
        topper_marks = mark
        topper_name = name

print("\nTopper:", topper_name)
print("Topper marks:", topper_marks)

print("\nPassed students:")
print(passed_students)
```

```python
Q4. Common Friends
--------------------------------------------------------------
Two friends have these friend lists on a social app:
  riya_friends  = {"Aman", "Neha", "Karan", "Sita", "Vikram"}
  arjun_friends = {"Neha", "Vikram", "Pooja", "Aman", "Rohan"}

Print:
  - Mutual friends of Riya and Arjun
  - Friends only Riya has
  - All people in either list (without duplicates)
  - Friend suggestions for Riya (Arjun's friends she doesn't know)
```

```python
riya_friends = {"Aman", "Neha", "Karan", "Sita", "Vikram"}

arjun_friends = {"Neha", "Vikram", "Pooja", "Aman", "Rohan"}

mutual_friends = riya_friends.intersection(arjun_friends)

riya_only = riya_friends.difference(arjun_friends)

all_friends = riya_friends.union(arjun_friends)

suggestions = arjun_friends.difference(riya_friends)

print("Mutual friends:", mutual_friends)
print("Friends only Riya has:", riya_only)
print("All people:", all_friends)
print("Friend suggestions for Riya:", suggestions)
```

```python
Q5. Character Counter
--------------------------------------------------------------
Take a word or sentence as input. Using a dictionary, count how
many times each character (ignoring spaces) appears. Then print
the character that appears the most.

Sample run:
  Enter text: banana
  {'b': 1, 'a': 3, 'n': 2}
  Most frequent character: a (3 times)
```

```python
text = input("Enter text: ")

character_count = {}

for char in text.lower():

    if char == " ":
        continue

    if char in character_count:
        character_count[char] = character_count[char] + 1
    else:
        character_count[char] = 1

print(character_count)

most_frequent = ""
highest_count = 0

for char, count in character_count.items():

    if count > highest_count:
        highest_count = count
        most_frequent = char

print("Most frequent character:", most_frequent,
      "(", highest_count, "times)")
```

```python
Q6. Restaurant Menu Billing
--------------------------------------------------------------
A restaurant menu is stored as a dictionary:
  menu = {"dosa": 60, "idli": 40, "vada": 30,
          "coffee": 25, "tea": 15}

Keep asking the customer what they want to order until they
type "bill". For each item:
  - If it is on the menu, add its price to the total
  - If not, print "Item not available"
At the end, print the total bill. If the total is more than
Rs. 200, apply a 10% discount.
```

```python
menu = {
    "dosa": 60,
    "idli": 40,
    "vada": 30,
    "coffee": 25,
    "tea": 15
}

total = 0

while True:

    item = input("Enter item (or type bill): ").lower()

    if item == "bill":
        break

    if item in menu:
        total = total + menu[item]
        print("Added", item, "-", menu[item])
    else:
        print("Item not available")

print("\nOriginal total: Rs.", total)

if total > 200:
    discount = total * 10 / 100
    final_total = total - discount

    print("Discount: Rs.", discount)
    print("Final bill: Rs.", final_total)
else:
    final_total = total
    print("No discount")

print("Total to pay: Rs.", final_total)
```

```python

Q7. City Coordinates (Tuples)
--------------------------------------------------------------
Store cities with their coordinates as tuples in a list:
  cities = [("Hyderabad", 17.38, 78.48),
            ("Mumbai", 19.07, 72.87),
            ("Chennai", 13.08, 80.27),
            ("Delhi", 28.61, 77.20)]

Write a program to:
  - Print each city with its latitude and longitude by
    unpacking the tuple in the for loop
  - Find the city that is farthest north (highest latitude)
  - Try to change the latitude of Mumbai inside a tuple and
    write (as a comment) what error you get and why
```

```python
cities = [
    ("Hyderabad", 17.38, 78.48),
    ("Mumbai", 19.07, 72.87),
    ("Chennai", 13.08, 80.27),
    ("Delhi", 28.61, 77.20)
]

print("City coordinates:")

for city, latitude, longitude in cities:
    print(city, "- Latitude:", latitude, "Longitude:", longitude)

northern_city = cities[0]

for city, latitude, longitude in cities:

    if latitude > northern_city[1]:
        northern_city = (city, latitude, longitude)

print("\nFarthest north:", northern_city[0])
print("Latitude:", northern_city[1])
```

```python
Q8. Library Book Tracker
--------------------------------------------------------------
A library stores book availability in a dictionary:
  books = {"Python Basics": 3, "Data Science 101": 0,
           "Machine Learning": 2, "SQL Guide": 1}
(The number is how many copies are available.)

Ask the user for a book name to borrow:
  - If the book doesn't exist, print "Book not found"
  - If copies are 0, print "Currently unavailable"
  - Otherwise, reduce the count by 1 and print
    "Book issued successfully"
Print the updated dictionary at the end.
```

```python
books = {
    "Python Basics": 3,
    "Data Science 101": 0,
    "Machine Learning": 2,
    "SQL Guide": 1
}

book_name = input("Enter book name to borrow: ")

if book_name not in books:
    print("Book not found")

elif books[book_name] == 0:
    print("Currently unavailable")

else:
    books[book_name] = books[book_name] - 1
    print("Book issued successfully")

print("Updated books:")
print(books)
```

```python
Q9. Group Employees by Department
--------------------------------------------------------------
Employee data is stored as a list of tuples (name, department):
  employees = [("Ravi", "IT"), ("Anjali", "HR"),
               ("Suresh", "IT"), ("Meena", "Finance"),
               ("Arun", "HR"), ("Divya", "IT")]

Create a dictionary where each department is a key and the
value is a list of employee names in that department.
Then print how many employees each department has.

Expected dictionary:
  {"IT": ["Ravi", "Suresh", "Divya"],
   "HR": ["Anjali", "Arun"],
   "Finance": ["Meena"]}
```

```python
employees = [
    ("Ravi", "IT"),
    ("Anjali", "HR"),
    ("Suresh", "IT"),
    ("Meena", "Finance"),
    ("Arun", "HR"),
    ("Divya", "IT")
]

department_employees = {}

for name, department in employees:

    if department not in department_employees:
        department_employees[department] = []

    department_employees[department].append(name)

print("Employees by department:")
print(department_employees)

print("\nEmployee count by department:")

for department, names in department_employees.items():
    print(department, ":", len(names))
```

```python
Q10. Quiz Answer Checker
--------------------------------------------------------------
The correct answers of a 5-question quiz are stored in a tuple:
  answer_key = ("B", "C", "A", "D", "B")

Ask the student to enter their 5 answers one by one. Compare
each with the answer key and:
  - Print which question numbers were correct and which wrong
  - Store wrong question numbers in a set
  - Print the final score out of 5 and a remark:
      5     : Excellent
      3 - 4 : Good
      0 - 2 : Needs improvement
```

```python
answer_key = ("B", "C", "A", "D", "B")

wrong_questions = set()
score = 0

for i in range(5):

    answer = input("Enter answer for Question " + str(i + 1) + ": ")
    answer = answer.upper()

    if answer == answer_key[i]:
        print("Question", i + 1, "is correct")
        score = score + 1
    else:
        print("Question", i + 1, "is wrong")
        wrong_questions.add(i + 1)

print("\nFinal score:", score, "out of 5")

if len(wrong_questions) > 0:
    print("Wrong question numbers:", wrong_questions)
else:
    print("No wrong answers")

if score == 5:
    print("Remark: Excellent")

elif score >= 3:
    print("Remark: Good")

else:
    print("Remark: Needs improvement")
```
